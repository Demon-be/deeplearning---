```python
import numpy as np
import pandas as pd

class NeuralNetwork:
    def __init__(self, input_size, hidden_size1, hidden_size2, output_size, learning_rate=0.01, num_iterations=1000):
        self.learning_rate = learning_rate
        self.num_iterations = num_iterations
        
        # 初始化权重和偏置
        self.weights1 = np.random.randn(input_size, hidden_size1) * 0.01  # 输入层到第一隐藏层
        self.bias1 = np.zeros((1, hidden_size1))  # 第一隐藏层偏置
        
        self.weights2 = np.random.randn(hidden_size1, hidden_size2) * 0.01  # 第一隐藏层到第二隐藏层
        self.bias2 = np.zeros((1, hidden_size2))  # 第二隐藏层偏置
        
        self.weights3 = np.random.randn(hidden_size2, output_size) * 0.01  # 第二隐藏层到输出层
        self.bias3 = np.zeros((1, output_size))  # 输出层偏置

    def sigmoid(self, z):
        return 1 / (1 + np.exp(-z))

    def sigmoid_derivative(self, z):
        return self.sigmoid(z) * (1 - self.sigmoid(z)) # 计算 Sigmoid 函数的导数，用于反向传播中的梯度计算

    def fit(self, X, y):
        num_samples = X.shape[0]
        
        for _ in range(self.num_iterations):
            # 前向传播
            # 第一隐藏层
            z1 = np.dot(X, self.weights1) + self.bias1  # (num_samples, hidden_size1)
            a1 = self.sigmoid(z1)  # (num_samples, hidden_size1)
            
            # 第二隐藏层
            z2 = np.dot(a1, self.weights2) + self.bias2  # (num_samples, hidden_size2)
            a2 = self.sigmoid(z2)  # (num_samples, hidden_size2)
            
            # 输出层
            z3 = np.dot(a2, self.weights3) + self.bias3  # (num_samples, output_size)
            a3 = self.sigmoid(z3)  # (num_samples, output_size)
            
            # 计算损失（二元交叉熵）
            loss = -np.mean(y * np.log(a3) + (1 - y) * np.log(1 - a3))
            
            # 反向传播
            # 输出层误差
            dloss_da3 = (a3 - y) / num_samples  # (num_samples, output_size)
            dloss_dz3 = dloss_da3 * self.sigmoid_derivative(z3)  # (num_samples, output_size)
            
            # 第二隐藏层误差
            dloss_da2 = np.dot(dloss_dz3, self.weights3.T)  # (num_samples, hidden_size2)
            dloss_dz2 = dloss_da2 * self.sigmoid_derivative(z2)  # (num_samples, hidden_size2)
            
            # 第一隐藏层误差
            dloss_da1 = np.dot(dloss_dz2, self.weights2.T)  # (num_samples, hidden_size1)
            dloss_dz1 = dloss_da1 * self.sigmoid_derivative(z1)  # (num_samples, hidden_size1)
            
            # 更新参数
            self.weights3 -= self.learning_rate * np.dot(a2.T, dloss_dz3)  # (hidden_size2, output_size)
            self.bias3 -= self.learning_rate * np.sum(dloss_dz3, axis=0, keepdims=True)  # (1, output_size)
            
            self.weights2 -= self.learning_rate * np.dot(a1.T, dloss_dz2)  # (hidden_size1, hidden_size2)
            self.bias2 -= self.learning_rate * np.sum(dloss_dz2, axis=0, keepdims=True)  # (1, hidden_size2)
            
            self.weights1 -= self.learning_rate * np.dot(X.T, dloss_dz1)  # (input_size, hidden_size1)
            self.bias1 -= self.learning_rate * np.sum(dloss_dz1, axis=0, keepdims=True)  # (1, hidden_size1)
            
            # 打印损失（可选）
            if _ % 100 == 0:
                print(f"Iteration {_}, Loss: {loss}")

    def predict(self, X):
        # 前向传播
        z1 = np.dot(X, self.weights1) + self.bias1
        a1 = self.sigmoid(z1)
        
        z2 = np.dot(a1, self.weights2) + self.bias2
        a2 = self.sigmoid(z2)
        
        z3 = np.dot(a2, self.weights3) + self.bias3
        a3 = self.sigmoid(z3)
        
        # 将概率转换为类别
        y_pred = [1 if i > 0.5 else 0 for i in a3]
        return np.array(y_pred)


# 示例数据
X = np.array([[1, 2], [2, 3], [3, 4], [4, 5]])  # 特征矩阵 (4, 2)
y = np.array([[0], [0], [1], [1]])  # 真实标签 (4, 1)

# 创建神经网络模型实例
input_size = X.shape[1]  # 输入层大小 (2)
hidden_size1 = 4  # 第一隐藏层大小 (4)
hidden_size2 = 3  # 第二隐藏层大小 (3)
output_size = 1  # 输出层大小 (1)

model = NeuralNetwork(input_size, hidden_size1, hidden_size2, output_size, learning_rate=0.01, num_iterations=1000)

# 训练模型
model.fit(X, y)

# 进行预测
predictions = model.predict(X)
print("预测结果:", predictions)
```

```python
import numpy as np

class NeuralNetwork:
    def __init__(self, input_size, hidden_size1, hidden_size2, output_size, learning_rate=0.01, num_iterations=1000):
        self.learning_rate = learning_rate
        self.num_iterations = num_iterations
        
        # 初始化权重和偏置
        self.weights1 = np.random.randn(input_size, hidden_size1) * 0.01  # 输入层到第一隐藏层
        self.bias1 = np.zeros((1, hidden_size1))  # 第一隐藏层偏置
        
        self.weights2 = np.random.randn(hidden_size1, hidden_size2) * 0.01  # 第一隐藏层到第二隐藏层
        self.bias2 = np.zeros((1, hidden_size2))  # 第二隐藏层偏置
        
        self.weights3 = np.random.randn(hidden_size2, output_size) * 0.01  # 第二隐藏层到输出层
        self.bias3 = np.zeros((1, output_size))  # 输出层偏置

    def tanh(self, z):
        return np.tanh(z)

    def tanh_derivative(self, z):
        return 1 - np.tanh(z) ** 2

    def softmax(self, z):
        exp_z = np.exp(z - np.max(z, axis=1, keepdims=True))  # 防止数值溢出
        return exp_z / np.sum(exp_z, axis=1, keepdims=True)

    def cross_entropy_loss(self, y_true, y_pred):
        m = y_true.shape[0]
        log_likelihood = -np.log(y_pred[range(m), y_true.argmax(axis=1)])
        loss = np.sum(log_likelihood) / m
        return loss

    def fit(self, X, y):
        num_samples = X.shape[0]
        
        for _ in range(self.num_iterations):
            # 前向传播
            # 第一隐藏层
            z1 = np.dot(X, self.weights1) + self.bias1  # (num_samples, hidden_size1)
            a1 = self.tanh(z1)  # (num_samples, hidden_size1)
            
            # 第二隐藏层
            z2 = np.dot(a1, self.weights2) + self.bias2  # (num_samples, hidden_size2)
            a2 = self.tanh(z2)  # (num_samples, hidden_size2)
            
            # 输出层
            z3 = np.dot(a2, self.weights3) + self.bias3  # (num_samples, output_size)
            a3 = self.softmax(z3)  # (num_samples, output_size)
            
            # 计算损失（交叉熵）
            loss = self.cross_entropy_loss(y, a3)
            
            # 反向传播
            # 输出层误差
            dloss_dz3 = a3 - y  # (num_samples, output_size)
            
            # 第二隐藏层误差
            dloss_da2 = np.dot(dloss_dz3, self.weights3.T)  # (num_samples, hidden_size2)
            dloss_dz2 = dloss_da2 * self.tanh_derivative(z2)  # (num_samples, hidden_size2)
            
            # 第一隐藏层误差
            dloss_da1 = np.dot(dloss_dz2, self.weights2.T)  # (num_samples, hidden_size1)
            dloss_dz1 = dloss_da1 * self.tanh_derivative(z1)  # (num_samples, hidden_size1)
            
            # 更新参数
            self.weights3 -= self.learning_rate * np.dot(a2.T, dloss_dz3)  # (hidden_size2, output_size)
            self.bias3 -= self.learning_rate * np.sum(dloss_dz3, axis=0, keepdims=True)  # (1, output_size)
            
            self.weights2 -= self.learning_rate * np.dot(a1.T, dloss_dz2)  # (hidden_size1, hidden_size2)
            self.bias2 -= self.learning_rate * np.sum(dloss_dz2, axis=0, keepdims=True)  # (1, hidden_size2)
            
            self.weights1 -= self.learning_rate * np.dot(X.T, dloss_dz1)  # (input_size, hidden_size1)
            self.bias1 -= self.learning_rate * np.sum(dloss_dz1, axis=0, keepdims=True)  # (1, hidden_size1)
            
            # 打印损失（可选）
            if _ % 100 == 0:
                print(f"Iteration {_}, Loss: {loss}")

    def predict(self, X):
        # 前向传播
        z1 = np.dot(X, self.weights1) + self.bias1
        a1 = self.tanh(z1)
        
        z2 = np.dot(a1, self.weights2) + self.bias2
        a2 = self.tanh(z2)
        
        z3 = np.dot(a2, self.weights3) + self.bias3
        a3 = self.softmax(z3)
        
        # 返回预测类别
        return np.argmax(a3, axis=1)


# 示例数据（多元分类）
np.random.seed(42)
X = np.random.randn(100, 4)  # 100个样本，4个特征
y = np.random.randint(0, 3, 100)  # 3个类别
y_one_hot = np.eye(3)[y]  # 将标签转换为one-hot编码

# 创建神经网络模型实例
input_size = X.shape[1]  # 输入层大小 (4)
hidden_size1 = 5  # 第一隐藏层大小 (5)
hidden_size2 = 4  # 第二隐藏层大小 (4)
output_size = 3  # 输出层大小 (3，对应3个类别)

model = NeuralNetwork(input_size, hidden_size1, hidden_size2, output_size, learning_rate=0.01, num_iterations=1000)

# 训练模型
model.fit(X, y_one_hot)

# 进行预测
predictions = model.predict(X)
print("预测结果:", predictions)
print("真实标签:", y)
```
