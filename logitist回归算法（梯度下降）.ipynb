```python
import numpy as np
import pandas as pd

x = np.random.randn(5,6)
w = np.random.randn(5,6)
b = 0.5

print(x)
print(w)
```

```python
Z = np.dot(w.T,x)+b
print(Z)

X = np.array([[1, 2], [2, 3], [3, 4], [4, 5]]) #特征矩阵，每一行代表一个样本，每一列代表一个特征
y = np.array([0, 0, 1, 1]) #真实标签向量
print(X)
print(y)
```

```python
class LogisticRegression:
    def __init__(self, learning_rate=0.01, num_iterations=1000):
        self.learning_rate = learning_rate
        self.num_iterations = num_iterations
        self.weights = None #weights 用于存储模型的权重，初始化为 None
        self.bias = None #bias 用于存储模型的偏置，初始化为 None

    def sigmoid(self, z):
        return 1 / (1 + np.exp(-z))

    def fit(self, X, y):
        num_samples, num_features = X.shape #X.shape 返回特征矩阵 X 的形状，是一个包含两个元素的元组，分别表示样本数量和特征数量
        self.weights = np.zeros(num_features) #np.zeros(num_features) 创建一个长度为 num_features 的全零数组，将其赋值给 self.weights，作为权重的初始值
        self.bias = 0 #将偏置 self.bias 初始化为 0

        for _ in range(self.num_iterations):
            linear_model = np.dot(X, self.weights) + self.bias #即Z = np.dot(w.T,x)+b
            y_pred = self.sigmoid(linear_model)

            dw = (1 / num_samples) * np.dot(X.T, (y_pred - y)) #计算损失函数关于权重的梯度，除以样本数量是为了求平均梯度
            db = (1 / num_samples) * np.sum(y_pred - y) #计算损失函数关于偏置的梯度，同样除以样本数量求平均梯度

            self.weights -= self.learning_rate * dw 
            self.bias -= self.learning_rate * db #计算损失函数关于偏置的梯度，同样除以样本数量求平均梯度

    def predict(self, X):
        linear_model = np.dot(X, self.weights) + self.bias
        y_pred = self.sigmoid(linear_model)
        y_pred_cls = [1 if i > 0.5 else 0 for i in y_pred] #使用列表推导式，将预测概率大于 0.5 的样本预测为 1，否则预测为 0
        return np.array(y_pred_cls) #将预测结果列表转换为 numpy 数组并返回

# 示例数据
X = np.array([[1, 2], [2, 3], [3, 4], [4, 5]]) #特征矩阵，每一行代表一个样本，每一列代表一个特征
y = np.array([0, 0, 1, 1]) #真实标签向量

# 创建逻辑回归模型实例
model = LogisticRegression(learning_rate=0.01, num_iterations=1000)

# 训练模型
model.fit(X, y)

# 进行预测
predictions = model.predict(X)
print("预测结果:", predictions)
```
